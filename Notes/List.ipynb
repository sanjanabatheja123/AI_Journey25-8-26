Remove Specified Item
The remove() method removes the specified item.

```python
thislist = ["apple", "banana", "cherry", "banana", "kiwi"]
thislist.remove("banana")
print(thislist)
```

Remove Specified Index
The pop() method removes the specified index.

.pop() removes an item and also returns its value

```python
thislist = ["apple", "banana", "cherry"]
thislist.pop(1)
print(thislist)
```

```python
# Remove the last item:

thislist = ["apple", "banana", "cherry"]
thislist.pop()
print(thislist)
```

```python
#The del keyword also removes the specified index:

# Example
# Remove the first item:

thislist = ["apple", "banana", "cherry"]
del thislist[0]
print(thislist)
# The del keyword can also delete the list completely.

# Example
# Delete the entire list:

thislist = ["apple", "banana", "cherry"]
del thislist
```

```python
books = ["Dragonsbane", "The Hobbit", "Wonder", "Jaws"]
del books[0:3]
```

For a more concise way to remove duplicates, you can also use a dictionary. A dictionary is a Python data type that represents an associative collection of unique keys that are mapped to values. Since dictionaries can’t contain duplicate keys, you can use the .fromkeys() method to remove copies from your list

```python
phone_numbers = ["54123", "54123", "54456", "54789", "54789", "54123"]

phone_numbers = list(dict.fromkeys(phone_numbers))

phone_numbers
['54123', '54456', '54789']
```

Check if an item exists (Returns True or False)Use the in operator. It is the cleanest, fastest, and most readable way to verify membership

```python
fruits = ["apple", "banana", "cherry"]

if "banana" in fruits:
    print("Found it!")
```

```python
fruits = ["apple", "banana", "cherry"]

try:
    position = fruits.index("cherry")
    print(f"Found at index: {position}")  # Output: 2
except ValueError:
    print("Item not in list")
```

```python
grades = ["A", "B", "A", "C", "A"]

# Get indices of all "A" grades
indices = [index for index, value in enumerate(grades) if value == "A"]
print(indices)  # Output: [0, 2, 4]
```
